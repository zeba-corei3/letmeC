```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-2*m.pi, 2*m.pi)
```

```python
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = 1/y3 #cot(x)
y5 = 1/y2 #sec(x)
y6 = 1/y1 #cosec(x)
```

```python
plt.plot(x, y1, "#FF0000", label="sinx")
plt.plot(x, y2, "#00FF00", label="cosx")
plt.plot(x, y3, "#0000FF", label="tanx")
plt.ylim(-10, 10)
plt.legend()
plt.show()
```

```python
plt.plot(x, y4, color="#FFFF00", marker="o", markersize=5, linewidth=0.2, label="cotx")
plt.plot(x, y5, color="#FF00FF", marker="o", markersize=5, linewidth=0.2, label="secx")
plt.plot(x, y6, color="#000000", marker="o", markersize=5, linewidth=0.2, label="cosecx")
plt.ylim(-10, 10)
plt.legend()
plt.show()
```

```python
p = np.linspace(-10, 10, 100)
```

```python
q1 = np.pow(p, 2)
q2 = np.pow(p, 3)
q3 = np.pow(p, 4)
q4 = np.pow(p, 5)
```

```python
plt.plot(p, p, "#000000", label="y=x")
plt.plot(p, q1, "#FF0000", label="$x^{2}$")
plt.plot(p, q2, "#00FF00", label="$x^{3}$")
plt.plot(p, q3, "#0000FF", label="$x^{4}$")
plt.plot(p, q4, "#FFFF00", label="$x^{5}$")
plt.ylim(-1, 1)
plt.xlim(-1, 1)
plt.legend()
plt.show()
```
